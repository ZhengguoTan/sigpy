```python
import numpy as np
import numpy.testing as npt
import random

import sigpy as sp

from sigpy.mri import retro
from sigpy.mri.dims import *
```

```python
kdat = np.zeros([1, 1, 1, 1, 1, 150, 100])
        
N_h = kdat.shape[DIM_ECHO]
N_y = kdat.shape[DIM_Y]
ETL = 35

nonzero_ky_ind_i = []
for h in range(N_h):
    ky_ind = random.sample(range(N_y), ETL)
    # print('> %2d: '%(h), ky_ind)

    kdat[:, :, h, ..., ky_ind, :] = 1.
    nonzero_ky_ind_i.append(ky_ind)

nonzero_ky_ind_i = np.array(nonzero_ky_ind_i)
print('> nonzero_ky_ind_i: ', nonzero_ky_ind_i.shape)

nonzero_ky_ind_o = retro.find_nonzero_ky_lines(kdat, ky_axis=DIM_Y)
print('> nonzero_ky_ind_o: ', nonzero_ky_ind_o.shape)
```
